```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Memuat data dari repository GitHub kamu agar bisa jalan mulus di Colab
csv_url = 'https://raw.githubusercontent.com/ZaxkyyOfficial/Flyrank-AI/main/data/raw/content_refresh_anonymized.csv'

try:
    file_path = '../../data/raw/content_refresh_anonymized.csv'
    df = pd.read_csv(file_path if os.path.exists(file_path) else csv_url)

    total_rows = len(df)

    if 'trend_direction' in df.columns:
        declining_pages = len(df[df['trend_direction'] == 'down'])
        base_rate = (declining_pages / total_rows) * 100

        print(f"1. Total units of analysis (web pages): {total_rows:,} rows.")
        print(f"2. Total pages with a downward trend (potential for refresh): {declining_pages:,} pages.")
        print(f"3. Base rate (random chance): {base_rate:.2f}%\n")
        print("This dataset is sufficient and suitable for analysis over the next 7 weeks.")
    else:
        print(f"Successfully loaded {total_rows:,} rows, but the 'trend_direction' column was not found.")

except Exception as e:
    print(f"Failed to load data: {e}")
```

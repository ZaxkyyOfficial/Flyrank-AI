## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*
Using static rules—such as "IF impressions_90d > 500 AND age_days > 180 THEN refresh"—results in an approach that is too rigid and overlooks the interplay between metrics.

Why ML is superior: Machine Learning models (such as Random Forest or Gradient Boosting) can automatically capture non-linear patterns and dynamic weighting. A model can learn that a 90-day-old page requires a refresh if its CTR has plummeted, whereas a 300-day-old page might remain viable if engagement levels are stable. ML yields far richer and more accurate probabilities for ranking purposes than the simple "Yes/No" flags produced by static rules.
